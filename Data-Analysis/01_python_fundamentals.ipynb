Expected file contents: `ALPHA`, `BETA`, `GAMMA`, each on its own line.


```python
from pathlib import Path

input_path = Path("practice_input.txt")
output_path = Path("practice_clean.txt")
cleaned_lines = []

for line in input_path.read_text(encoding="utf-8").splitlines():
    stripped_line = line.strip()
    if stripped_line:
        cleaned_lines.append(stripped_line.upper())

output_path.write_text("\n".join(cleaned_lines), encoding="utf-8")
print(output_path.read_text(encoding="utf-8"))
```

For a non-negative integer duration, return `(hours, remaining_minutes)`. Unpack one result into two variables.

### 8.4

Return `(total, average)` for a list of numbers. For an empty list, return `(0, None)`.


```python
def summarise(values):
    total = sum(values)
    if not values:
        return total, None
    return total, total / len(values)


print(summarise([2, 0, 4]))
print(summarise([]))
```

### 8.5

Return a sorted list of unique values in `required` that are absent from `available`. Use sets.


```python
def find_missing(required, available):
    return sorted(set(required) - set(available))


print(find_missing([3, 1, 3, 5], [1, 2]))
print(find_missing([], [1]))
```

### 8.6

Complete the helpers and dispatcher:

- `difference(a, b)`: return `a - b`.
- `percentage(a, b)`: return `a / b * 100`, or None if `b == 0`.
- `calculate(a, b, operation)`: call the matching helper for `"difference"` or `"percentage"`; return None for other operations or non-numeric inputs. Assume booleans are not supplied.


```python
def difference(a, b):
    return a - b


def percentage(a, b):
    if b == 0:
        return None
    return a / b * 100


def calculate(a, b, operation):
    if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
        return None
    if operation == "difference":
        return difference(a, b)
    if operation == "percentage":
        return percentage(a, b)
    return None


print(calculate(15, 20, "difference"))
print(calculate(15, 20, "percentage"))
print(calculate(1, 0, "percentage"))
print(calculate("1", 2, "difference"))
print(calculate(1, 2, "other"))
```
