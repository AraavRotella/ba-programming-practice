```python
import pandas as pd
import numpy as np

def check(label, got_fn, expected, tol=1e-3):
    try:
        got = got_fn()
        if isinstance(expected, float) and not isinstance(got, (list, dict, pd.Series, pd.DataFrame)):
            ok = abs(float(got) - expected) < tol
        else:
            ok = got == expected
            if hasattr(ok, "all"):
                ok = bool(ok.all())
    except Exception as e:
        got, ok = f"error: {type(e).__name__}: {e}", False
    print(("\u2705 " if ok else "\u274c ") + label + ("" if ok else f"\n   got:      {got!r}\n   expected: {expected!r}"))
```
